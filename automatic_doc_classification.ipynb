```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline

documents = [
    "Python and Java are programming languages",
    "Machine learning uses artificial intelligence",
    "Cricket and football are popular sports",
    "Players participate in sports tournaments"
]

labels = ["Technology", "AI", "Sports", "Sports"]

model = make_pipeline(TfidfVectorizer(), MultinomialNB())
model.fit(documents, labels)

test = ["Artificial intelligence and machine learning"]

prediction = model.predict(test)

print("Document Category:", prediction[0])
```
